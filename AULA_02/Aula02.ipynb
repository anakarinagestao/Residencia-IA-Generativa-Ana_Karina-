```python
from docling.document_converter import DocumentConverter

converter = DocumentConverter()

arquivos_pdf = [
    "bioetica_e_ia.pdf",
    "escrita_academica_ia.pdf",
    "twitter_algoritmo.pdf"
]

for pdf in arquivos_pdf:
    resultado = converter.convert(pdf)
    markdown_texto = resultado.document.export_to_markdown()

    nome_md = pdf.replace(".pdf", ".md")
    with open(nome_md, "w", encoding="utf-8") as f:
        f.write(markdown_texto)

    print(f"✅ Convertido: {pdf} → {nome_md}")
```

```python
from google.colab import userdata
from openai import OpenAI

client = OpenAI(
    api_key=userdata.get('OPENAI_API_KEY'),
    base_url="https://openrouter.ai/api/v1"
)

modelo = "openai/gpt-oss-20b:free"
```

```python
def extrair_metadados(caminho_md):
    with open(caminho_md, "r", encoding="utf-8") as f:
        conteudo = f.read()

    response = client.chat.completions.create(
        model=modelo,
        messages=[
            {
                "role": "system",
                "content": "Você extrai metadados de artigos acadêmicos. Responda APENAS com um JSON válido usando EXATAMENTE estas chaves em português: 'titulo', 'autores', 'ano'. Não traduza os nomes das chaves."
            },
            {
                "role": "user",
                "content": f"Extraia o título, autores e ano de publicação do texto abaixo. Use as chaves 'titulo', 'autores' e 'ano' no JSON de resposta:\n\n{conteudo[:4000]}"
            }
        ],
        response_format={
            "type": "json_schema",
            "json_schema": {
                "name": "metadados_artigo",
                "strict": True,
                "schema": {
                    "type": "object",
                    "properties": {
                        "titulo": {"type": "string"},
                        "autores": {
                            "type": "array",
                            "items": {"type": "string"}
                        },
                        "ano": {"type": "integer"}
                    },
                    "required": ["titulo", "autores", "ano"],
                    "additionalProperties": False
                }
            }
        }
    )

    resultado = json.loads(response.choices[0].message.content)
    return resultado
```

```python
arquivos_md = [
    "aula_2/bioetica_e_ia.md",
    "aula_2/escrita_academica_ia.md",
    "aula_2/twitter_algoritmo.md"
]

for md in arquivos_md:
    print(f"Processando: {md}")
    metadados = extrair_metadados(md)
    print(json.dumps(metadados, indent=2, ensure_ascii=False))

    nome_json = md.replace(".md", ".json")
    with open(nome_json, "w", encoding="utf-8") as f:
        json.dump(metadados, f, indent=2, ensure_ascii=False)

    print(f"✅ Salvo: {nome_json}\n")
```
