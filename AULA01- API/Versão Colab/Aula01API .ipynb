# Introdução à IA - Primeiro Contato com LLMs
Neste notebook, vamos fazer a nossa primeira chamada para a API da OpenAI utilizando as chaves configuradas com segurança.


```python
from openai import OpenAI
from google.colab import userdata

client = OpenAI(
    base_url="https://openrouter.ai/api/v1",
    api_key=userdata.get('OPENROUTER_API_KEY')
)
```

### 2. Chamada à API
Usaremos o modelo configurado por padrão (`gpt-4o-mini`) para gerar um texto.


```python
modelo = "nvidia/nemotron-3-ultra-550b-a55b:free"

response = client.chat.completions.create(
    model=modelo,
    messages=[
        {"role": "user", "content": "write a haiku about ai"}
    ],
)

print(response.choices[0].message.content)
```
